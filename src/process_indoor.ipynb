### image alignment (optional)

align nir_rgb rectified image pairs

```python
import os
import glob
import numpy as np
import cv2
from alignR2L import align
folders = ['sec1','sec2','sec3','sec4','sec5_1','sec6','sec7','sec8']
for folder in folders:
  path = '/media/jiayu/9395-FCAC/neucs2_outdoor/outdoor/processed/train/' + folder + '/nir_rgb'

  left_paths = glob.glob(path+'/left/*.png')
  os.makedirs(path+'/r2l', exist_ok=True)
  for left_file in left_paths:
    right_file = left_file.replace('left', 'right')
    disp_file = left_file.replace('left', 'label_cre_unfiltered').replace('png', 'npy')
    R2L = np.uint8(align(left_file, right_file, disp_file))
    cv2.imwrite(left_file.replace('left','r2l'),R2L)
```
